# numpy的常用函数

```python
import numpy as np
```

## 1. 基本数学函数

```python
# 平方根
print(np.sqrt(9))
print(np.sqrt([1,4,9]))
```

```python
# 指数(e^X)
print(np.exp(1))
```

```python
# 自然对数(lnX)
print(np.log(1))
print(np.log(np.exp(1)))
```

```python
# 三角函数
print(np.sin(np.pi/2))
print(np.cos(np.pi/2))
```

```python
# 绝对值
print(np.abs(-1))
print(np.abs([-1,2,-3]))
```

```python
# a的b次幂
print(np.power(2,3))
print(np.power([2,3,4],2))
```

```python
# 四舍五入
# .5四舍五入时要确保最终结果为偶数
print(np.round(1.5))
print(np.round([1.5,2.3,3.7]))
```

```python
# 向上/向下取整
print(np.ceil(1.5)) # up
print(np.floor(1.5)) # down

```

```python
# 检测缺失值(NaN)
print(np.isnan(np.nan))
print(np.isnan(np.nan))
print(np.isnan([1,2,3,np.nan]))
print(np.isnan([1,2,3,np.nan,4]))
```

## 2. 统计常用函数
求和，计算平均数、中位数、标准差、方差
查找最大、最小值
计算分位数、累计和、累计差

```python
arr = np.random.randint(1,20,8)
print(arr)
```

```python
# 求和
print(np.sum(arr))
```

```python
# 平均值
print(np.mean(arr))
print(np.average(arr))
```

```python
# 中位数
print(np.median(arr))
# 上下四分位数
print(np.percentile(arr,25))
print(np.percentile(arr,75))
```

```python
# 方差、标准差
print(np.var(arr))
print(np.std(arr))
```

```python
# 查找最大、最小值，以及他们对应的索引值
print(np.max(arr),np.argmax(arr))
print(np.min(arr),np.argmin(arr))
```

```python
# 累计和、累积积
arr = np.array([1,2,3])
print(np.cumsum(arr))
print(np.cumprod(arr))
```

## 2. 比较函数
比较函数是否大于、等于、小于
逻辑与或非
检查数组中是否有一个True，是否所有的数都为True

```python
# 是否大于
print(np.greater([1,2,3],[2,2,2]))
# 是否小于
print(np.less([1,2,3],[2,2,2]))
# 是否大于等于
print(np.greater_equal([1,2,3],[2,2,2]))
# 是否小于等于
print(np.less_equal([1,2,3],[2,2,2]))
# 逻辑与或非
print(np.logical_and([True,False],[True,True]))
print(np.logical_or([True,False],[True,True]))
print(np.logical_not([True,False]))
# 检查数组中是否有一个True
print(np.any([True,False]))
# 检查数组中是否所有的数都为True
print(np.all([True,True]))
```

```python
#自定义条件
arr = np.random.randint(0,100,15)
print(arr)
# np.where(条件,符合条件的，不符合条件的),可以嵌套使用
# eg. 二值化
print(np.where(arr >= 60, arr, 0))
print(np.where(arr>=60,1,0))
```

```python
# np.select(条件(list)，返回结果(list))
print(np.select([arr<60,arr>=60],['不及格','及格'],default='null'))
```
