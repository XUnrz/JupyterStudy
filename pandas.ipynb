# pandas

## series的创建

```python
import pandas as pd
```

```python
# series的创建
s = pd.Series([1,2,3,4,5])
# 索引 值 dtype
# 自定义索引值
s = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
# 自定义标签
s = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'], name='s')
print(s)
```

```python
# 通过字典创建
s = pd.Series({'a':1, 'b':2, 'c':3})
print(s)
s1 = pd.Series(s,index=['a','c'])
print(s1)
```

## Series的属性
index,values,dtype(s),shape,ndim,size,name 部分类似于numpy <br>
loc[] 显式索引 <br>
iloc[] 隐式索引 <br>
at[] 通过标签访问单个元素 <br>
iat[] 通过位置访问单个元素 <br>


```python
# 基本属性
s = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'], name='s')
print(s)
# index:索引对象
print(s.index)
# values:ndarray对象
print(s.values)
# dtype:数据类型
print(s.dtype)
# shape:形状
print(s.shape)
# ndim:维度
print(s.ndim)
# size:大小
print(s.size)
# name:名称
print(s.name)
```

```python
# 索引方式
# 这两个均支持切片
# 显式索引
print(s.loc['a']) # 类似于字典查找方式
# 隐式索引
print(s.iloc[0]) # 类似于列表查找方式
# at[] (不支持切片，用的方法和显式相同)
# iat[] (不支持切片，用的方法和隐式相同)
```

```python
# 访问数据
s = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'], name='s')
# print(s[1]) 貌似新版不支持
print(s['a'])
```

```python
# 布尔索引
s = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'], name='s')
print(s[s < 3])
print(s[s % 2 == 0])
```

## series的常用方法
head() # 默认会返回前五行数据<br>
tail() # 默认返回后五行数据<br>
isin() # 是否在集合中<br>
isna() # 是否为NaN<br>
sum() # 求和<br>
mean() # 平均值<br>
min() # 最小值<br>
max() # 最大值<br>
var() # 方差<br>
std() # 标准差<br>
median() # 中位数<br>
mode() # 众数<br>
quantile() # 分位数<br>
describe() # 描述统计<br>

```python
# 常用方法
s = pd.Series([1,2,3,4,5,6,7],index=['a','b','c','d','e','f','g'],name='s')
print(s.head()) # 默认会返回前五行数据
print(s.tail()) # 默认会返回后五行数据
print(s.isin([1,3,5]))
print(s.isna())
print(s.sum())
print(s.mean())
print(s.min())
print(s.max())
print(s.var())
print(s.std())
print(s.median())
print(s.mode())
print(s.quantile(0.5))
print(s.describe())
```

## 常用方法2
value_counts() # 统计唯一值出现的次数<br>
count() # 统计非NaN的个数<br>
unique() # 获取去重后数组<br>
nunique() # 返回唯一值的个数（去重）<br>
drop_duplicates() # 删除重复值<br>
sample() # 随机抽样<br>
sort_index() # 按索引排序<br>
sort_values() # 按值排序<br>
replace() # 替换值<br>
keys() # 返回索引<br>

```python
# 常用方法2
s = pd.Series([1,2,3,4,5,6,7,7,7,6],index=['a','b','c','d','e','f','g','h','i','j'],name='s')
print(s.value_counts())
print(s.count())
print(s.unique())
print(s.nunique())
print(s.drop_duplicates())
print(s.sample(3))
print(s.sort_index())
print(s.sort_values())
print(s.replace(1,10))
print(s.keys())
```
